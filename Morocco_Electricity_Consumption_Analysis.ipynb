### Data Quality Assessment

The dataset was assessed for duplicate records, duplicate timestamps, date-time formatting and recording interval consistency.

The duplicate record check returned zero results, indicating that no identical rows were identified. Similarly, the duplicate timestamp check returned zero results, confirming that each observation has a unique timestamp.

The `Datetime` column was successfully converted from an object data type into a datetime format, enabling time-based analysis.

The recording interval assessment confirmed that all 52,415 consecutive time differences were exactly 10 minutes, indicating consistent recording frequency.

The dataset begins on 1 January 2017 at 00:00 and ends on 30 December 2017 at 23:50. Although the dataset represents the 2017 reporting period, the final calendar day, 31 December 2017, is not included. This limitation will be considered when interpreting monthly and annual consumption patterns.

Overall, the dataset has passed the initial structural and timestamp quality checks and is suitable for further preparation and exploratory analysis.

```
# Extract year
df['Year'] = df['Datetime'].dt.year

# Extract month number
df['Month'] = df['Datetime'].dt.month

# Extract month name
df['Month_Name'] = df['Datetime'].dt.month_name()

# Extract day
df['Day'] = df['Datetime'].dt.day

# Extract hour
df['Hour'] = df['Datetime'].dt.hour

# Extract day of the week
df['Day_of_Week'] = df['Datetime'].dt.day_name()

# Extract date only
df['Date'] = df['Datetime'].dt.date

print("Time-based columns created successfully!")
```

```
measurement_columns = [
    'Temperature',
    'Humidity',
    'WindSpeed',
    'GeneralDiffuseFlows',
    'DiffuseFlows',
    'PowerConsumption_Zone1',
    'PowerConsumption_Zone2',
    'PowerConsumption_Zone3'
]

outlier_results = []

for column in measurement_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_results.append({
        'Column': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower_Bound': lower_bound,
        'Upper_Bound': upper_bound,
        'Outlier_Count': len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_results)

outlier_summary
```

```
# Identify Zone 3 outliers

Q1 = df['PowerConsumption_Zone3'].quantile(0.25)
Q3 = df['PowerConsumption_Zone3'].quantile(0.75)

IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR

zone3_outliers = df[
    df['PowerConsumption_Zone3'] > upper_bound
].copy()

# Extract month and hour

zone3_outliers['Month'] = zone3_outliers['Datetime'].dt.month
zone3_outliers['Hour'] = zone3_outliers['Datetime'].dt.hour

# Count outliers by month

print("Zone 3 Outliers by Month:")
print(zone3_outliers.groupby('Month').size())

# Count outliers by hour

print("\nZone 3 Outliers by Hour:")
print(zone3_outliers.groupby('Hour').size().sort_values(ascending=False))
```

```
# Compare normal readings against outliers

normal_consumption = df[
    df['PowerConsumption_Zone3'] <= upper_bound
]['PowerConsumption_Zone3']

outlier_consumption = zone3_outliers[
    'PowerConsumption_Zone3'
]

print("Normal Consumption Average:",
      round(normal_consumption.mean(), 2))

print("Outlier Consumption Average:",
      round(outlier_consumption.mean(), 2))

print("Maximum Outlier Consumption:",
      round(outlier_consumption.max(), 2))

print("Percentage Difference:",
      round(
          ((outlier_consumption.mean() -
            normal_consumption.mean()) /
           normal_consumption.mean()) * 100, 2
      ), "%")
```

```
print("Normal vs Outlier Weather Conditions")

weather_comparison = pd.DataFrame({
    'Normal': df.loc[
        df['PowerConsumption_Zone3'] <= upper_bound,
        ['Temperature', 'Humidity', 'WindSpeed']
    ].mean(),

    'Outliers': zone3_outliers[
        ['Temperature', 'Humidity', 'WindSpeed']
    ].mean()
})

print(weather_comparison.round(2))
```
